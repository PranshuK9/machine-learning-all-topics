```python
from sklearn.datasets import make_regression
import numpy as np
import pandas as pd
from sklearn.model_selection import cross_val_score
```

```python
X,y = make_regression(n_samples=4, n_features=1, n_targets=1, noise=80, random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
reg.fit(X,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X, reg.predict(X),color='red')
```

```python
#Lets apply gradient descent assuming slope is constant = 78.35
#and lets assume the starting value for intercept b = 0
y_pred = ((78.35*X)+0).reshape(4)
```

```python
y_pred
```

```python
plt.scatter(X,y)
plt.plot(X, reg.predict(X),color='red',label='OLS')
plt.plot(X, y_pred,color='blue',label='b=0')
plt.legend()
```

```python
m = 78.35
b = 0
loss_slope = -2 * np.sum(y- m*X.ravel() - b)
loss_slope
```

```python
#lets take learning rate = 0.1
lr = 0.1
step_size = loss_slope*lr
step_size
```

```python
#Calculating the new intercept
b = b - step_size
b
```

```python
y_pred1 = ((78.35*X)+b).reshape(4)

plt.scatter(X,y)
plt.plot(X, reg.predict(X),color='red',label='OLS')
plt.plot(X, y_pred1,color='green',label='b={}'.format(b))
plt.plot(X, y_pred,color='blue',label='b=0')
plt.legend()
```

```python
#Iteration 2
loss_slope = -2 * np.sum(y- m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((78.35*X)+b).reshape(4)

plt.scatter(X,y)
plt.plot(X, reg.predict(X),color='red',label='OLS')
plt.plot(X, y_pred2,color='brown',label='b={}'.format(b))
plt.plot(X, y_pred1,color='green',label='b={}'.format(b))
plt.plot(X, y_pred,color='blue',label='b=0')
plt.legend()
```

```python
#Iteration 3
loss_slope = -2 * np.sum(y- m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope*lr
step_size
```

```python
b = b - step_size
b
```
